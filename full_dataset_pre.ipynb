## 0. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os

from sklearn.preprocessing import OneHotEncoder
```

## 1. Read Dataset as DataFrame --> "business+economy.csv"

```python
cwd = os.getcwd()

folder_name = "data"

# Dataset contains flight information and prices
file_name = "business+economy.csv"

file_p = os.path.join(cwd, folder_name, file_name)

data = pd.read_csv(file_p)
```

#### Drop Unrequired Feature Columns

```python
# First column --> Column counting for ticket number
data = data.iloc[:, 1:]

# Column of flight's code --> Not important
data = data.drop('flight', axis=1)
```

## 3. One-Hot Encoding

```python
categorical_features_label = data.select_dtypes(include=['object']).columns.to_list()

print(categorical_features_label)
```

```python
encoder = OneHotEncoder(sparse_output=False)

one_hot_encoded = encoder.fit_transform(data[categorical_features_label])

one_hot_data = pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(categorical_features_label))

data_encoded = pd.concat([data, one_hot_data], axis=1)

data_encoded = data_encoded.drop(categorical_features_label, axis=1)
```

```python
data_encoded.info()
```

```python
data_encoded.describe()
```

## 4. Save Cleaned Dataset (No Category) AS CSV

```python
folder_name = "data"

file_name = "full_dataset_cleaned_noCategory.csv"

file_final_path = os.path.join(os.getcwd(), folder_name, file_name)

if os.path.exists(file_final_path):
    print(f"\nFile already exists. \nPlease check at the filepath: {file_final_path}\n")

else:
    data_encoded.to_csv(file_final_path, header=True, index=False)
    print(f"\nFile sucessfully saved at {file_final_path}\n")
```
